```python
import numpy as np
import matplotlib.pyplot as plt

print("Keplerian ellipses for e = 0.0 ... 0.9 (separate plots)")
print("Units: a = 1 AU, GM = 4π² AU³/yr², time in years")

# Fixed orbital parameters
a = 1.0
GM = 4.0 * np.pi**2
n = np.sqrt(GM / a**3)  # 2π / yr

# Newton–Raphson solver for Kepler's equation M = E - e sin E
def solve_kepler(M, e, tol=1e-10, max_iter=20):
    if e == 0.0:
        return M
    E = M
    for _ in range(max_iter):
        f = E - e * np.sin(E) - M
        if abs(f) < tol:
            break
        fp = 1.0 - e * np.cos(E)
        E = E - f / fp
    return E
es = np.arange(0.0, 1.0, 0.1)
M_vals = np.linspace(0.0, 2.0*np.pi, 600)

# 10 distinct colors
colors = ['b', 'g', 'r', 'c', 'm', 'y', 'k',
          'tab:orange', 'tab:purple', 'tab:brown']
for idx, e in enumerate(es):
    color = colors[idx % len(colors)]

    # Solve Kepler for each M
    E_vals = np.array([solve_kepler(M, e) for M in M_vals])
    r_vals = a * (1.0 - e * np.cos(E_vals))
    nu_vals = 2.0 * np.arctan2(
        np.sqrt(1 + e) * np.sin(E_vals/2.0),
        np.sqrt(1 - e) * np.cos(E_vals/2.0)
    )

    # Cartesian coordinates
    x = r_vals * np.cos(nu_vals)
    y = r_vals * np.sin(nu_vals)

    # (plotting added in next commit)
    print(f"Computed orbit for e = {e:.1f}")
```

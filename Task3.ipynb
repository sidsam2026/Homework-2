```python
import numpy as np
import matplotlib.pyplot as plt

print("Keplerian ellipses for e = 0.0 ... 0.9 (separate plots)")
print("Units: a = 1 AU, GM = 4π² AU³/yr², time in years")

a = 1.0
GM = 4.0 * np.pi**2
n = np.sqrt(GM / a**3)  # 2π / yr

def solve_kepler(M, e, tol=1e-10, max_iter=20000):
    if e == 0.0:
        return M
    E = M
    for _ in range(max_iter):
        f = E - e * np.sin(E) - M
        if abs(f) < tol:
            break
        fp = 1.0 - e * np.cos(E)
        E = E - f / fp
    return E
es = np.arange(0.0, 1.0, 0.1)
M_vals = np.linspace(0.0, 2.0*np.pi, 600)
```
